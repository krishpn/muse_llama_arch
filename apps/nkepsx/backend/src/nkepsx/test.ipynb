```python
import pandas as pd

excel_file= pd.ExcelFile("~/Downloads/Ashadh-2078-2.xlsx", engine='openpyxl')
excel_file= pd.ExcelFile("~/Downloads/Shrawan_2083_Publish.xlsx", engine='openpyxl')
sheet_names = excel_file.sheet_names

print(f"sheets{sheet_names}")

all_sheets = pd.read_excel(excel_file, sheet_name=None)

for sheet_name, df in all_sheets.items():
    print(f"\n" + "="*50)
    print(f"SHEET NAME: {sheet_name}")
    print("="*50)
    
    # 1. Drop rows and columns that are entirely NaN (the layout noise)
    df_clean = df.dropna(how='all').dropna(axis=1, how='all')
    
    # 2. Check for remaining null values
    null_counts = df_clean.isnull().sum()
    has_nulls = null_counts.sum() > 0
    
    print(f"Has Null/NaN values: {has_nulls}")
    if has_nulls:
        print("Columns with missing values:")
        print(null_counts[null_counts > 0])
    
    print("\n--- Content Preview ---")
    if df_clean.empty:
        print("[Sheet is empty after cleaning]")
    else:
        # Print the first few rows of the cleaned data
        print(df_clean.head(3))

```

```python
import pdfplumber
import pandas as pd
import requests
import io

# URL of the PDF file
pdf_url = "https://www.nrb.org.np/contents/uploads/2026/09/Shrawan_2083_Publish.pdf"

print("Downloading and opening PDF from URL...")
response = requests.get(pdf_url)
pdf_file = io.BytesIO(response.content)

with pdfplumber.open(pdf_file) as pdf:
    for page_num, page in enumerate(pdf.pages, start=1):
        print(f"\n" + "="*50)
        print(f"--- PAGE {page_num} ---")
        print("="*50)
        
        # Extract tables from the page
        tables = page.extract_tables()
        
        if not tables:
            print("[No structured tables found on this page]")
            continue
            
        for table_idx, table in enumerate(tables):
            print(f"\nTable {table_idx + 1}:")
            
            # Convert the raw list-of-lists table to a pandas DataFrame
            df = pd.DataFrame(table)
            
            # Clean up rows and columns where everything is None or empty strings
            df = df.dropna(how='all').dropna(axis=1, how='all')
            
            if df.empty:
                print("[Table is empty after cleaning]")
            else:
                # Print the first few rows of this table
                print(df.head(5).to_string(index=False, header=False))
```
